```python
from autogluon.tabular import TabularDataset, TabularPredictor
import pandas as pd
```

```python
data = TabularDataset('diabetes.csv')
```

```python
predictor = TabularPredictor(label='Outcome').fit(data,num_gpus=1)
```

```python
test = pd.read_csv('diabetes.csv')
test.drop(columns=['Outcome'])
```

```python
predictor.evaluate(test, silent=True)
```

```python
predictor.leaderboard(test)
```
